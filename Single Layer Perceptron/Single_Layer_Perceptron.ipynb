```python
import numpy as np
```

```python
class Perceptron:
  def __init__(self,num_inputs,learning_rate=0.1):
    self.weights=np.random.uniform(-1,1,num_inputs+1)
    self.learning_rate=learning_rate

  def linear(self,inputs):
    return np.dot(inputs,self.weights[1:])+self.weights[0]

  def heaviside_step_fn(self,z):
    if z>=0:
      return 1
    else:
      return 0

  def predict(self, inputs):
    if np.ndim(inputs) == 1:
        return self.heaviside_step_fn(self.linear(inputs))
    outputs = []
    for x in inputs:
        outputs.append(self.heaviside_step_fn(self.linear(x)))
    return np.array(outputs)

  def loss(self,prediction,target):
     return target-prediction

  def train(self,inputs,target):
    prediction=self.predict(inputs)
    error=self.loss(prediction,target)
    self.weights[1:]+=self.learning_rate*error*inputs
    self.weights[0]+=self.learning_rate*error

  def fit(self,X,y,epochs=20):
   for epoch in range(epochs):
    for input,target in zip(X,y):
      self.train(input,target)

  def accuracy(self,X,y):
   predictions=self.predict(X)
   return np.mean(predictions==y)*100
```

```python
#and
X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])
y = np.array([0,0,0,1])
model=Perceptron(2,0.1)
model.fit(X,y,20)

print("AND prediction:", model.predict(X))
print("AND accuracy:", model.accuracy(X,y))
```
